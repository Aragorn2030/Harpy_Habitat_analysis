```python
import os
import numpy as np
import pandas as pd
import geopandas as gpd
import folium
from folium.plugins import HeatMap
from sklearn.cluster import DBSCAN
```

```python
# 1. Load data (GBIF export is tab-separated)
df = pd.read_csv("Harpia harpyja.csv", sep="\t", low_memory=False)
df
```

```python
# 2. Clean data
df = df[df["occurrenceStatus"] == "PRESENT"]
df = df[~df["basisOfRecord"].isin(["FOSSIL_SPECIMEN", "MATERIAL_SAMPLE", "MATERIAL_CITATION"])]
df = df.dropna(subset=["decimalLatitude", "decimalLongitude"])
df = df[~((df["decimalLatitude"] == 0) & (df["decimalLongitude"] == 0))]
df = df[df["decimalLatitude"].between(-35, 25) & df["decimalLongitude"].between(-100, -30)]
df = df.drop_duplicates(subset=["decimalLatitude", "decimalLongitude", "eventDate"])
df = df.reset_index(drop=True)
print("Clean records:", len(df))
```

```python
# 3. CRS: create point geometries in WGS84 (EPSG:4326)
gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df["decimalLongitude"], df["decimalLatitude"]),
    crs="EPSG:4326",
)
gdf
```

```python
# 4. Hotspots: DBSCAN with 10 km radius and at least 10 records
coords = np.radians(gdf[["decimalLatitude", "decimalLongitude"]].values)
gdf["cluster"] = DBSCAN(eps=10 / 6371, min_samples=10, metric="haversine").fit(coords).labels_
```

```python
# 6. Map
m = folium.Map(location=[gdf.geometry.y.mean(), gdf.geometry.x.mean()], zoom_start=5)
pa = None 
if pa is not None:
    folium.GeoJson(pa, name="Protected areas").add_to(m)

HeatMap(gdf[["decimalLatitude", "decimalLongitude"]].values.tolist(), name="Density").add_to(m)

for _, g in gdf[gdf["cluster"] != -1].groupby("cluster"):
    folium.Marker(
        [g["decimalLatitude"].mean(), g["decimalLongitude"].mean()],
        tooltip=f"{len(g)} records",
    ).add_to(m)

folium.LayerControl().add_to(m)
m.save("harpia_map.html")
m
```
